```python
import sys
from pathlib import Path
import re
import string

import nltk
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords

import spacy

# Download necessary NLTK data
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')

# Load spacy model
try:
    nlp = spacy.load('en_core_web_sm')
except:
    import spacy.cli
    spacy.cli.download('en_core_web_sm')
    nlp = spacy.load('en_core_web_sm')

print("Setup Complete.")
```

## 3. Dataset / Corpus Input
Define the sample text corpora for preprocessing operations.

```python
text_21 = """Natural Language Processing (NLP) is a branch of Artificial Intelligence! 
It helps computers understand and process human language in 2026. 
NLP is used in 100+ applications such as chatbots, machine translation, sentiment analysis, and text summarization. 

Text preprocessing is an important step in NLP. It includes several operations such as cleaning text, removing punctuation, and removing unnecessary numbers (e.g., 1, 2, 3). 
For example, a chatbot may process 500+ customer messages every day. These messages can contain CAPITAL LETTERS, numbers like 404, punctuation marks (.,!?), and    extra spaces. 
The main goal of text preprocessing is to convert raw text into a clean and consistent format. This makes the text easier to analyze and use for further natural language processing tasks."""

text_22 = """Natural Language Processing (NLP) is a fascinating field of Artificial Intelligence. It enables computers to understand human language. Tokenization is essential!"""

text_23 = """Natural Language Processing is an important field of Artificial Intelligence. It is used in many applications such as chatbots and machine translation. Some of these are very helpful to us."""

print("Text corpora loaded.")
```

## 4. Text Preprocessing & Linguistic Normalization
### 4.1 Basic Text Preprocessing

```python
# 1. Lowercase
upper_count = sum(1 for c in text_21 if c.isupper())
lowercase_text = text_21.lower()

# 2. Punctuation
punct_found = [c for c in text_21 if c in string.punctuation]
punct_types = set(punct_found)
no_punct_text = text_21.translate(str.maketrans('', '', string.punctuation))

# 3. Numbers
numbers_found = re.findall(r'\d', text_21)
no_nums_text = re.sub(r'\d', '', no_punct_text)

# 4. Whitespace
extra_whitespace = len(re.findall(r'\s{2,}', text_21))
clean_text = " ".join(no_nums_text.split())

print(f"Uppercase count: {upper_count}")
print(f"Different punctuation types found ({len(punct_types)}): {punct_types}")
print(f"Numbers found: {len(numbers_found)}")
print(f"Extra whitespace instances: {extra_whitespace}")
print("\nCleaned Text:")
print(clean_text.lower())
```

### 4.2 Tokenization

```python
print("--- Method 1: Without NLTK/spaCy ---")
words_simple = text_22.split()
sents_simple = re.split(r'(?<=[.!?])\s+', text_22)
print(f"Words: {words_simple[:5]}...")
print(f"Sentences: {sents_simple[:2]}...")

print("\n--- Method 2: NLTK ---")
words_nltk = word_tokenize(text_22)
sents_nltk = sent_tokenize(text_22)
print(f"Words: {words_nltk[:5]}...")
print(f"Sentences: {sents_nltk[:2]}...")

print("\n--- Method 3: spaCy ---")
doc = nlp(text_22)
words_spacy = [token.text for token in doc]
sents_spacy = [sent.text for sent in doc.sents]
print(f"Words: {words_spacy[:5]}...")
print(f"Sentences: {sents_spacy[:2]}...")
```

### 4.3 Stop Word Removal

```python
stop_words = set(stopwords.words('english'))

tokens = word_tokenize(text_23)
filtered_tokens = [w for w in tokens if not w.lower() in stop_words]
found_stops = [w for w in tokens if w.lower() in stop_words]

print(f"Original Token Count: {len(tokens)}")
print(f"Filtered Token Count: {len(filtered_tokens)}")
print("\nStop Words Found in Text:")
print(set(found_stops))
print("\nFiltered Tokens:")
print(filtered_tokens[:10])
```
