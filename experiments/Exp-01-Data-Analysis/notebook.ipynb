# Experiment 1: Data Analysis

**Course:** Computational Linguistics and Natural Language Processing Lab  
**Dataset:** `data/employee_information_100.csv`

### Load Libraries and Dataset

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("data/employee_information_100.csv")
df.head()
```

### Experiment 1.1: Average Salary of Each Department

```python
df.groupby("Department")["Salary"].mean().plot(kind="bar", title="Average Salary by Department", ylabel="Salary")
plt.show()
```

### Experiment 1.2: Number of Employees in Each Department

```python
df["Department"].value_counts()
```

### Experiment 1.3: Percentage of Male and Female Employees

```python
df["Gender"].value_counts().plot(kind="pie", autopct="%.1f%%", title="Gender Distribution")
plt.ylabel("")
plt.show()
```

### Experiment 1.4: Salary Distribution

```python
df["Salary"].plot(kind="hist", bins=10, title="Salary Distribution", edgecolor="black")
plt.xlabel("Salary")
plt.show()
```

### Experiment 1.5: Relationship Between Experience and Salary

```python
df.plot(kind="scatter", x="Experience_Years", y="Salary", title="Experience vs Salary")
plt.show()
```

### Experiment 1.6: Top 10 Employees with Highest Salary

```python
df.nlargest(10, "Salary")[["Employee_ID", "Name", "Department", "Salary"]]
```

### Experiment 1.7: Highest Salary in Every Department

```python
df.groupby("Department")["Salary"].max()
```

### Experiment 1.8: Employees with Salary Greater than Overall Average

```python
df[df["Salary"] > df["Salary"].mean()][["Employee_ID", "Name", "Department", "Salary"]]
```

### Experiment 1.9: Average Years of Experience in Each Department

```python
df.groupby("Department")["Experience_Years"].mean()
```

### Experiment 1.10: Distribution of Employee Ages

```python
df["Age"].plot(kind="hist", bins=10, title="Age Distribution", edgecolor="black")
plt.xlabel("Age")
plt.show()
```
